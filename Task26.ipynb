```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

### `Q 1-3` Using the 'diamonds' dataset - sns.load_dataset('diamonds')
1. Create a violinplot of 'price' grouped by 'cut'.
2. Create regplot on `carat` vs `price`. and give hue on 'cut'
3. Create boxplot on 'color' and 'price'

```python
# code here

df = sns.load_dataset("diamonds")

sns.violinplot(x="cut", y="price", data=df)
plt.title("Price Distribution by Cut")
plt.show()

sns.lmplot(x="carat", y="price", hue="cut", data=df)
plt.title("Carat vs Price by Cut")
plt.show()

sns.boxplot(x="color", y="price", data=df)
plt.title("Price by Color")
plt.show()
```

### `Q 4` Using the 'Taxis' dataset - sns.load_dataset('taxis')
4.1. Create a categorical estimate plot of the totl fare - 'total' for each payment type - 'payment'.

4.2. Create a regression plot on time of ride  vs total fare. You will need to calculate ride time using pickup and dropoff column.

4.3 Give hue on payment type. and Another plot hue on taxi 'color'. Observe the plot.

```python
# code here

df = sns.load_dataset("taxis")

sns.barplot(x="payment", y="total", data=df)
plt.title("Total Fare by Payment Type")
plt.show()

df["pickup"] = pd.to_datetime(df["pickup"])
df["dropoff"] = pd.to_datetime(df["dropoff"])
df["ride_time"] = (df["dropoff"] - df["pickup"]).dt.total_seconds() / 60

sns.lmplot(x="ride_time", y="total", hue="payment", data=df)
plt.title("Ride Time vs Total Fare (by Payment)")
plt.show()

sns.lmplot(x="ride_time", y="total", hue="color", data=df)
plt.title("Ride Time vs Total Fare (by Taxi Color)")
plt.show()
```

```python
# code here

url = "https://docs.google.com/spreadsheets/d/e/2PACX-1vQVpcVtdYdZU4zU4-lqxt-iPHkyndDWs_aqEDUu9ZodlJ48Dku0PFgdXlj2N5RCmwXJrNtZLsI_wEVf/pub?gid=220677750&single=true&output=csv"
df = pd.read_csv(url)
```

```python
plt.figure(figsize=(10,5))
sns.stripplot(x="gender", y="bloodpressure", hue="smoker", data=df, dodge=True)
plt.title("Strip Plot: Blood Pressure by Gender and Smoker")
plt.show()

plt.figure(figsize=(10,5))
sns.swarmplot(x="gender", y="bloodpressure", hue="smoker", data=df)
plt.title("Swarm Plot: Blood Pressure by Gender and Smoker")
plt.show()
```

### **`Problem 6:`** Draw a Box-plot and a Violin plot of which x-axis represents the "region" and the y-axis represents the "bmi". Also add extra information of the column "diabetic".

```python
# code here

plt.figure(figsize=(10,5))
sns.boxplot(x="region", y="bmi", hue="diabetic", data=df)
plt.title("Box Plot: BMI by Region with Diabetic Status")
plt.show()

plt.figure(figsize=(10,5))
sns.violinplot(x="region", y="bmi", hue="diabetic", data=df, split=True)
plt.title("Violin Plot: BMI by Region with Diabetic Status")
plt.show()
```

### **`Problem  7:`** Draw a bar plot and point plot of which x-axis represents the "gender" and y-axis represents "claim". Also add extra information about "smoker" column.

```python
# code here

plt.figure(figsize=(10,5))
sns.barplot(x="gender", y="claim", hue="smoker", data=df)
plt.title("Bar Plot: Claim by Gender with Smoker")
plt.show()

plt.figure(figsize=(10,5))
sns.pointplot(x="gender", y="claim", hue="smoker", data=df)
plt.title("Point Plot: Claim by Gender with Smoker")
plt.show()
```

### **`Problem 8:`** Draw a reg plot between "age" and "bmi" columns.

### **`Problem 9:`** Draw a pair plot of the insurance data. Use "gender" as hue parameter.

```python
# code here

sns.pairplot(df, hue="gender")
plt.suptitle("Pair Plot of Insurance Data", y=1.02)
plt.show()
```

### **`Problem 10:`** Draw a pair grid of the insurance data and use "diabetic" column as a hue parameter. Also, make the diagonal plots as box-plot, upper parts as scatter plot and the lower parts as kde plot.

```python
# code here

num_df = df.select_dtypes(include="number")

g = sns.PairGrid(num_df.join(df["diabetic"]), hue="diabetic")

g.map_upper(sns.scatterplot)
g.map_lower(sns.scatterplot)
g.map_diag(sns.histplot)

g.add_legend()
plt.show()
```

### **`Prolem 11:`** Draw a joint plot as scatter between "bloodpressure" and "bmi". Use "smoker" as hue parameter.

```python
# code here

sns.jointplot(data=df, x="bloodpressure", y="bmi", hue="smoker", kind="scatter")
plt.show()
```

### **`Problem 12:`** Draw a joint grid of which x-axis represents "age" and y-axis represents "claim". Draw

```python
# code here

g = sns.JointGrid(data=df, x="age", y="claim")

g.plot_joint(sns.scatterplot)
g.plot_marginals(sns.histplot)

g.fig.suptitle("Joint Grid: Age vs Claim", y=1.02)

plt.show()
```
