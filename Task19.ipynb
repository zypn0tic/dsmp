### `Q-1:` Using `groupby` make groups using the `"Pclass"` column and find out the average age and total number of missing values in the `"Age"` column for every group.

````python
# code here

result = titanic_df.groupby('Pclass')['Age'].agg(Average_Age = 'mean', Missing_Values=lambda x: x.isna().sum())

print(result)
````

### `Q-2:` Using `groupby` make groups using the `"Pclass"` column and fill every group's `"Embarked"` column's missing values with the mode value of that group. After that, print every group's `"Embarked"` column's value counts in ascending order.

````python
# code here

titanic_df['Embarked'] = titanic_df.groupby('Pclass')['Embarked'].transform(lambda x: x.fillna(x.mode()[0]))

result = titanic_df.groupby('Pclass')['Embarked'].value_counts().sort_values()
print(result)
````

### `Q-3:` Make groups based on `"Embarked"` column. And for each of this embarked group, make another group based on `"Pclass"` and find out the average fare (round off up to 2 decimal places) for each "Pclass" for each group of "Embarked".

**Sample Output:**

```bash
{'C': {1: 105, 2: 25, 3: 11},
 'Q': {1: 90, 2: 12, 3: 11},
 'S': {1: 70, 2: 20, 3: 15}}
```

````python
# code here

grouped = titanic_df.groupby(['Embarked', 'Pclass'])['Fare'].mean().round(2)

result = grouped.unstack().to_dict(orient='index')

print(result)
````

### `Q-4:` Perform `groupby` based on the `"Team"` column and then perform Z Normalization on top of the below columns of each group:
1. Passes
2. Passes Completed
3. Attempted Line Breaks
4. Completed Line Breaks

You have to make a python function named `z_normalization` which takes two arguments:

1. *group:* Every group that you have created
2. *cols_to_perform:* This parameter takes a list of columns on which you have to perform the Z-Normalization.

For this problem, you have to use th `apply()` method.

$\Large Z \ - \ Normalization = \frac{X_i - \mu}{std} $

After that find out the below values for each group:
- minimum "Passess"
- maximum "Passess"
- minimum "Yellow Cards"
- maximum "Yellow cards"
- average "Yellow Cards"
- maximum "Attempted Line Breaks"
- minimum "Attempted Line Breaks"
- standardard deviation of "Attempted Line Breaks"
- average Possession

````python
# code here

def z_normalization(group, cols_to_perform):
    group = group.copy()
    for col in cols_to_perform:
        mean = group[col].mean()
        std = group[col].std()
        group[col] = (group[col] - mean) / std
    return group

cols = [
    'Passes',
    'Passes Completed',
    'Attempted Line Breaks',
    'Completed Line Breaks'
]

normalized_df = fifa_df.groupby('Team').apply(
    lambda g: z_normalization(g, cols)
).reset_index(drop=True)

result = normalized_df.groupby('Team').agg({
    'Passes': ['min', 'max'],
    'Yellow Cards': ['min', 'max', 'mean'],
    'Attempted Line Breaks': ['max', 'min', 'std'],
    'Possession (%)': 'mean'
})

print(result)
````

### `Q-5:` Find batsman in below category-
* Highest score while chasing
* Best Strike rate while chasing and have faced 100+ balls


> Chasing mean team batting in second inning

````python
# code here

df = pd.read_csv("datasets/ipl_deliveries.csv")
df.sample(5)
df.info()
````

````python
chasing = df[df['innings'] == 2]

# highest score while chasing
scores = chasing.groupby(['ID', 'batter'])['batsman_run'].sum()
print("Highest score while chasing: ", scores.idxmax()[1], scores.max())

bat_stats = chasing.groupby('batter').agg({
    'batsman_run': 'sum', 
    'ballnumber': 'count'
})

bat_stats.rename(columns={'ballnumber':'balls'}, inplace=True)
bat_stats = bat_stats[bat_stats['balls'] >= 100]
bat_stats['SR'] = (bat_stats['batsman_run'] / bat_stats['balls']) * 100

print("Best strike rate while chasing: ", bat_stats['SR'].idxmax(), bat_stats['SR'].max())
````

### `Q-6` Most Successful bowler against any batsman. Find that pair of bowler and batsman.
> Most Successful in terms of dissmissal. A bowler who have dissmissed any batsman most no of times. If any two pairs have same no of dissmisal, consider runs conceded by bowler to that batsman. Those who have concede lesser runs is more successful.

````python
# code here

dismissals = df[(df['isWicketDelivery'] == 1) & (df['player_out'] == df['batter'])]
wickets = dismissals.groupby(['bowler','batter']).size().reset_index(name='dismissals')
runs = df.groupby(['bowler','batter'])['batsman_run'].sum().reset_index(name='runs')

stats = wickets.merge(runs, on=['bowler','batter'])
stats = stats.sort_values(['dismissals','runs'], ascending=[False, True])

print(stats.iloc[0])
````

### `Q-7`: Most successful batting pair in IPL. Batting pair who have scored most runs playing together.

````python
# code here
df['pair'] = df.apply(
    lambda x: tuple(sorted([x['batter'], x['non-striker']])),
    axis=1
)

pair_runs = df.groupby('pair')['total_run'].sum()

print("Most successful batting pair:", pair_runs.idxmax())
print("Total runs together:", pair_runs.max())
````

### `Q-8:` Make a dataframe for all batting pairs played together.
```
Batsman1 Batsman2 Runs Avg StrikeRate
```

> Just to ease this question you can count wide-balls for strike rate.

````python
# code here

df['Batsman1'], df['Batsman2'] = zip(
    *df.apply(lambda x: sorted([x['batter'], x['non-striker']]), axis=1)
)

pair_stats = df.groupby(['Batsman1','Batsman2']).agg(
    Runs=('total_run','sum'),
    Balls=('ballnumber','count')
).reset_index()

pair_stats['Avg'] = pair_stats['Runs'] / pair_stats['Balls']
pair_stats['StrikeRate'] = (pair_stats['Runs'] / pair_stats['Balls']) * 100

pair_df = pair_stats[['Batsman1','Batsman2','Runs','Avg','StrikeRate']]

print(pair_df)
````
