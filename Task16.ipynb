### `Q-2:` Write a Pandas program to add, subtract, multiple and divide two Pandas Series.

````python
# code here

# create two Series
s1 = pd.Series([10, 20, 30, 40])
s2 = pd.Series([1, 2, 3, 4])

# addition
print("Addition:")
print(s1 + s2)

# subtraction
print("\nSubtraction:")
print(s1 - s2)

# multiplication
print("\nMultiplication:")
print(s1 * s2)

# division
print("\nDivision:")
print(s1 / s2)
````

### `Q-5.`Write a function to change the data type of given a column or a Series. Function takes series and data type as input, returns the converted series.
```
series = pd.Series([1,2,'Python', 2.0, True, 100])
change to float type data
```
Note: Read about pd.to_numeric()

````python
# code here

def change_dtype(series, dtype):
    if dtype == "float":
        return pd.to_numeric(series, errors="coerce").astype(float)
    else:
        return series.astype(dtype)

series = pd.Series([1,2,'Python',2.0,True,100])
converted = change_dtype(series, "float")
print(converted)
````

### `Q-9`
    i. Read `items.csv` making `item_name` as index.
    ii. Show no of nan values
    ii. Item price is given in $, so convert it to rupees without currency symbol.
    iii. Make data type of newly made series as float.
    iv. Fill nan with mean of the series


How csv file looks

```
item_name	item_price
Chips and Fresh Tomato Salsa	$2.39
Izze	$3.39
Nantucket Nectar	$3.39
Chips and Tomatillo-Green Chili Salsa	$2.39
Chicken Bowl	$16.98

```

````python
# i. Read `items.csv` making `item_name` as index.
df = pd.read_csv("datasets/items.csv", index_col="item_name").squeeze()
print(df.head())
````

````python
# ii. Show no of nan values

print(df.isna().sum())
````

````python
# iii. Item price is given in $, so convert it to rupees without currency symbol.

df = df.replace(r'\$', '', regex=True).astype(float)

df = df * 143
print(df.head())
````

````python
# iv. Fill nan with mean of the series

df = df.fillna(df.mean())
print(df)
````

### `Q-10`:
    i. Find mean price
    ii. Find 30th and 6th percentile value
    iii. No of items price lies between [1000 to 2000]
    iv. Plot Histogram on price with bin size 50

````python
# code here

# Mean price
print(df.mean())

# Percentiles
print(df.quantile(0.30))
print(df.quantile(0.06))

# Items between 1000 and 2000
print(df[(df >= 1000) & (df <= 2000)].count())

# Histogram
import matplotlib.pyplot as plt
plt.hist(df, bins=50)
plt.show()
````
