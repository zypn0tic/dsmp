### `Q-1:` Create a random 3x4 matrix with value between 0-100. And perform below tasks

    i. Sort this matrix. np.sort()
    ii. Sort this matrix based on values in 2nd column.
    iii. Sort this matrix based on max value in each row.
    iv. Sort based on elements value.


See examples:
```
arr =
    [[92 90 74]
    [ 6 63 93]
    [15 93 96]
    [70 60 48]]
```
```
i. np.sort
[[74 90 92]
 [ 6 63 93]
 [15 93 96]
 [48 60 70]]

ii. based on 2nd column
[[70 60 48]
 [ 6 63 93]
 [92 90 74]
 [15 93 96]]

iii. based on row max- ascending
[[15 93 96]
 [ 6 63 93]
 [92 90 74]
 [70 60 48]]

iv. based on elements value
[[ 6 15 48]
 [60 63 70]
 [74 90 92]
 [93 93 96]]

```

````python
# code here

arr = np.array([[92, 90, 74],
      [ 6, 63, 93],
      [15, 93, 96],
      [70, 60, 48]])

# i.sortin'

print(np.sort(arr))
````

````python
# ii. based on 2nd column 

select_col = arr[:,1]
sorted_arr = arr[select_col.argsort()]

print(sorted_arr)
````

````python
# iii. based on row max-ascending 

row_max = arr.max(axis=1)

sorted_arr = arr[row_max.argsort()[::-1]]
print(sorted_arr)
````

````python
# iv. based on elements value 

sorted_elements = np.sort(arr, axis=None).reshape(arr.shape)

print(sorted_elements)
````

### `Q-2:` There is an array of marks of 5 students in 4 subjects. Further you are asked to perform below task.
    i. Add marks every student of an extra subject in the same array.
    ii. Add two new students marks in respective 5 subjects.(one subject added in above task)
    iii. Add extra column with sum of all subjects(5-subjects) marks
    iv. Sort the array(non-ascending order) on total marks column--one added in above task. Show top 2 rows.

Note: Change dimension of arrays during concatenation or appending if required.

Given Array-
```
marks = [[13, 10,  9, 33],
       [63, 46, 90, 42],
       [39, 76, 13, 29],
       [82,  9, 29, 78],
       [67, 61, 59, 36]]

extra_subject = [41, 87, 72, 36, 92]
# Two extra students record-
rec1 = [77, 83, 98, 95, 89]
rec2 = [92, 71, 52, 61, 53]
```

````python
# code here

marks = np.array([
    [13, 10,  9, 33],
    [63, 46, 90, 42],
    [39, 76, 13, 29],
    [82,  9, 29, 78],
    [67, 61, 59, 36]
])

extra_subject = np.array([41, 87, 72, 36, 92])
rec1 = np.array([77, 83, 98, 95, 89])
rec2 = np.array([92, 71, 52, 61, 53])

# i) Add extra subject (as a new column)
marks = np.hstack((marks, extra_subject.reshape(-1, 1)))

# ii) Add two new students (as new rows)
marks = np.vstack((marks, rec1, rec2))

# iii) Add total marks column (sum of 5 subjects)
totals = marks.sum(axis=1).reshape(-1, 1)
marks = np.hstack((marks, totals))

# iv) Sort by total marks descending and show top 2 rows
sorted_marks = marks[marks[:, -1].argsort()[::-1]]
top2 = sorted_marks[:2]

print("Final array with totals:\n", marks)
print("\nTop 2 students by total marks:\n", top2)
````

### `Q-3:` Find unique arrays from a 2D array column wise and row wise.
```
arr = np.array([[1,2,3,3,1,1],
                [0,9,1,2,8,8],
                [1,2,3,8,8,8],
                [1,2,3,3,1,1]])
```
Expected Result-
```
Row Wise
[[0 9 1 2 8 8]
 [1 2 3 3 1 1]
 [1 2 3 8 8 8]]

Col Wise
[[1 1 2 3 3]
 [0 8 9 1 2]
 [1 8 2 3 8]
 [1 1 2 3 3]]
```

````python
# code here

arr = np.array([[1,2,3,3,1,1],
                [0,9,1,2,8,8],
                [1,2,3,8,8,8],
                [1,2,3,3,1,1]])

# Row-wise unique
row_unique = np.unique(arr, axis=0)

# Column-wise unique
col_unique = np.unique(arr.T, axis=0).T

print("Row Wise")
print(row_unique)

print("\nCol Wise")
print(col_unique)
````

### `Q-5:` Get row numbers of NumPy array having element larger than X.
```
arr = [[1,2,3,4,5],
      [10,-3,30,4,5],
      [3,2,5,-4,5],
      [9,7,3,6,5]]

X = 6
```

````python
# code here

arr = np.array([[1,2,3,4,5],
                [10,-3,30,4,5],
                [3,2,5,-4,5],
                [9,7,3,6,5]])

X = 6

rows = np.where(np.any(arr > X, axis=1))[0]

print("Row indices:", rows)
````

### `Q-7:` You are given a array. You have to find the minimum and maximum array element and remove that from the array.

```python
import numpy as np

np.random.seed(400)
arr = np.random.randint(100, 1000, 200).reshape((1, 200))
```

````python
# code here

np.random.seed(400)
arr = np.random.randint(100, 1000, 200).reshape((1, 200))

# find min and max
min_val = arr.min()
max_val = arr.max()

# flatten array
arr_flat = arr.flatten()

# remove min and max
new_arr = arr_flat[(arr_flat != min_val) & (arr_flat != max_val)]

print("Minimum:", min_val)
print("Maximum:", max_val)
print("Array after removing min & max:")
print(new_arr)
````

### `Q-8:` You are given an arrays. You have to limit this array's elements between 100 to 200. $arr ∈ [100, 700]$. So replace those values accordingly with the minimum and maximum value. Then sort the array and perform the cumulative sum of that array.

````python
# code here

np.random.seed(400)
arr = np.random.randint(100, 700, 20)

print("Original array:")
print(arr)

# limit values between 100 and 200
arr_limited = np.clip(arr, 100, 200)

# sort array
sorted_arr = np.sort(arr_limited)

# cumulative sum
cumsum_arr = np.cumsum(sorted_arr)

print("\nLimited array:")
print(arr_limited)

print("\nSorted array:")
print(sorted_arr)

print("\nCumulative sum:")
print(cumsum_arr)
````

### `Q-9:` You are given a array ($arr ∈ [0, 1]$). First you have round off the elements upto 3 decimal places and compare that
- 0th percentile == minimum value of the array
- 100th percentile == maximum value of the array
- also find the difference betwen 51th percenile and 50th percentile values

````python
# code here

np.random.seed(400)
arr = np.random.random(20)

# round to 3 decimal places
arr = np.round(arr, 3)

# percentiles
p0 = np.percentile(arr, 0)
p100 = np.percentile(arr, 100)

p50 = np.percentile(arr, 50)
p51 = np.percentile(arr, 51)

difference = p51 - p50

print("Array:", arr)
print("\n0th Percentile:", p0)
print("Minimum:", arr.min())

print("\n100th Percentile:", p100)
print("Maximum:", arr.max())

print("\nDifference between 51st and 50th percentile:", difference)
````
