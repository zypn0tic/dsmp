Pandas Series are like a column in a table. 1D array holding any data type.

```python
import numpy as np
import pandas as pd
```

Using List 

```python
Mountains = ['Everest', 'K2', 'Kangchenjunga', 'Lhotse', 'Makalu', 'Nuptse', 'Pumori', 'Tawachengmaa']
Heights = [8848, 8611, 8586, 8516, 8485, 8382, 8365, 8363]

pd.Series(Heights, index=Mountains, name='Mountains', dtype=float)
```

Using Dictionary

```python
phone = {
    "Ram": 981234567,
    'Hari': 987123456,
    'Gita' : 987654321
}

phone_series = pd.Series(phone, name = 'Phone Numbers')
```

phone is an Object of series class. so it can access the attribute and Methods of that (Series) Class 

#### Series Attributes

```python
# Size
print(phone_series.size)

# Index
print(phone_series.index)

# Values
print(phone_series.values)
print(type(phone_series.values))

# Type
print(phone_series.dtype)

# Head
print(phone_series.head(2))

# Tail
print(phone_series.tail(2))

# is_unique
print(phone_series.is_unique)
```

#### Series Math Methods 

```python
print(bollywood.count(), "\n") # Count the number of non-null values in each column

print(bollywood.isnull().sum(), "\n") # Count the number of null values in each column

print(bollywood.isnull().sum().sum(), "\n") # Count the number of null values in the entire dataframe

print(bollywood.isnull().sum().sum() / bollywood.size * 100, "\n") # Count the percentage of null values in the entire dataframe

print(bollywood.isnull().sum().sum() / bollywood.size * 100, "\n") # Count the percentage of null values in the entire dataframe
```

```python
# Works only on numerical columns

print(bollywood.sum(), end="\n\n") # Sum of all the values in the column

print(bollywood.mean(), end="\n\n") # Mean of all the values in the column

print(bollywood.median(), end="\n\n") # Median of all the values in the column

print(bollywood.mode(), end="\n\n") # Mode of all the values in the column

print(bollywood.std(), end="\n\n") # Standard deviation of all the values in the column

print(bollywood.var(), end="\n\n") # Variance of all the values in the column
```

```python
print(bollywood.min(), end="\n\n") # Minimum value in the column
print(bollywood.max(), end="\n\n") # Maximum value in the column
```

#### Series Indexing

```python
x = pd.Series([1, 2, 3, 4, 5])
print(x[2])

# print(x[-1]) # Negative indexing doesnot work in pandas series
```

```python
bollywood[bollywood.rating > 8]
```

```python
y = pd.Series([1, 2, 3, 4, 5])

# Slicing
print(y[1:4], end="\n\n")

# Slicing with step
print(y[1:4:2])

# Slicing with negative index
print(y[-4:-1], end="\n\n")

# Slicing with negative index and step
print(y[-4:-1:2], end="\n\n")

# Slicing with negative index and step
print(y[-4:-1:2], end="\n\n")

# Fancy Indexing
print(y[[0, 2, 4]], end="\n\n")

```

```python
# Editing in Series
x = pd.Series([1, 2, 3, 4, 5])

# Editing in Series
x[2] = 10
print(x, end="\n\n")

# Editing in Series
x[1:4] = 20
print(x, end="\n\n")
```

#### Series with Python Functionalities

```python
# len/type/dir/sorted/min/max

print(len(bollywood), end="\n\n")
print(type(bollywood), end="\n\n")
print(dir(bollywood), end="\n\n")
print(sorted(bollywood), end="\n\n")
print(min(bollywood), end="\n\n")
print(max(bollywood), end="\n\n")
```

```python
# Type Conversion
x = pd.Series([1, 2, 3, 4, 5])
print(x, end="\n\n")

x = x.astype(str)
print(x, end="\n\n")

x = x.astype(int)
print(x, end="\n\n")

x = x.astype(float)

print(list(bollywood), end="\n\n")
print(dict(bollywood), end="\n\n")
```

```python
# Arthemetic Operations
x = pd.Series([1, 2, 3, 4, 5])
y = pd.Series([6, 7, 8, 9, 10])

print(x + y, end="\n\n")
print(x - y, end="\n\n")
print(x * y, end="\n\n")
print(x / y, end="\n\n")
print(x % y, end="\n\n")
print(x ** y, end="\n\n")

# Comparison Operators
print(x == y, end="\n\n")
print(x != y, end="\n\n")
print(x > y, end="\n\n")

# Relational Operators
print(x < y, end="\n\n")
print(x <= y, end="\n\n")
print(x >= y, end="\n\n")

# Logical Operators
print(x & y, end="\n\n")
print(x | y, end="\n\n")
print(x ^ y, end="\n\n")
```

```python
import matplotlib.pyplot as plt

# Create figure for histogram
plt.figure(figsize=(12, 6))

# Create histogram of ratings
plt.hist(bollywood['rating'], bins=20, edgecolor='black')

# Customize the plot
plt.title('Distribution of Bollywood Movie Ratings', fontsize=14, pad=15)
plt.xlabel('Rating (out of 10)', fontsize=12)
plt.ylabel('Number of Movies', fontsize=12)

# Add grid for better readability
plt.grid(True, alpha=0.3)

# Adjust layout
plt.tight_layout()

plt.show()

# Create figure for top movies
plt.figure(figsize=(12, 6))

# Get top 10 movies by rating (excluding 0 ratings which likely mean unrated)
top_movies = bollywood[bollywood['rating'] > 0].nlargest(10, 'rating')

# Create bar plot
plt.bar(range(len(top_movies)), top_movies['rating'])

# Set the movie titles as x-tick labels
plt.xticks(range(len(top_movies)), top_movies.index, rotation=45, ha='right')

# Customize the plot
plt.title('Top 10 Highest Rated Bollywood Movies', fontsize=14, pad=15)
plt.xlabel('Movie Title', fontsize=12)
plt.ylabel('Rating (out of 10)', fontsize=12)

# Add grid for y-axis
plt.grid(True, axis='y', alpha=0.3)

# Set y-axis to start from 0 and end at 10 (max rating)
plt.ylim(0, 10.5)

# Adjust layout to prevent label cutoff
plt.tight_layout()

plt.show()
```
